```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]

plt.plot(days, temperature,marker='o', linestyle = '--')

plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Temperature During the Week")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]

student1 = [50, 55, 60, 70, 75]
student2 = [45, 60, 65, 68, 80]

plt.plot(months, student1, marker="o", label="Student 1")
plt.plot(months, student2, marker="o", label="Student 2")

plt.xlabel("Month")
plt.ylabel("Marks")
plt.title("Student Performance")

plt.legend()

plt.show()
```

```python
hours = [1, 2, 3, 4, 5]
marks = [40, 45, 55, 70, 80]

plt.scatter(hours, marks)

plt.xlabel("Hours Studied")
plt.ylabel("Marks")

plt.show()
```

```python
marks = [45, 50, 52, 55, 60, 62, 65, 70, 72, 75, 80, 85]

plt.hist(marks)

plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
fig, ax = plt.subplots()

# ax.plot(x, y)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("My Graph")

plt.show()
```

```python
fig, ax = plt.subplots(1, 2)

ax[0].plot([1, 2, 3], [4, 5, 6])
ax[1].plot([1, 2, 3], [7, 8, 9])
ax[0].set_title("Left Graph")
ax[1].set_title("Right Graph")

plt.show()
```

```python
!pip install ucimlrepo

import pandas as pd
import matplotlib.pyplot as plt

from ucimlrepo import fetch_ucirepo

student_performance = fetch_ucirepo(id=320)

X = student_performance.data.features
y = student_performance.data.targets

data = pd.concat([X, y], axis=1)
```

```python
important_columns = [
    "studytime",
    "failures",
    "absences",
    "G1",
    "G2",
    "G3"
]

data[important_columns].head()
```

```python
fig, ax = plt.subplots()

ax.hist(data["G3"], bins=10)

ax.set_title("Distribution of Final Grades")
ax.set_xlabel("Final Grade (G3)")
ax.set_ylabel("Number of Students")

plt.show()
```

```python
fig, ax = plt.subplots()

ax.scatter(
    data["absences"],
    data["G3"]
)

ax.set_title("Absences vs Final Grade")
ax.set_xlabel("Number of Absences")
ax.set_ylabel("Final Grade")

plt.show()
```

```python
g1 = data["G1"].mean()
g2 = data["G2"].mean()
g3 = data["G3"].mean()

print(g1)
print(g2)
print(g3)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# First subplot
ax[0].hist(data["G3"], bins=20)

ax[0].set_title("Final Grade Distribution")
ax[0].set_xlabel("Final Grade")
ax[0].set_ylabel("Number of Students")


# Second subplot
ax[1].scatter(
    data["absences"],
    data["G3"]
)

ax[1].set_title("Absences vs Final Grade")
ax[1].set_xlabel("Absences")
ax[1].set_ylabel("Final Grade")


plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Seaborn ships a built-in copy of Iris — same data as the UCI version,
# already cleaned into a tidy DataFrame. No download/path needed.
iris = sns.load_dataset("iris")
print(iris.head())
```

```python
iris.head()
```

```python
plt.scatter(iris["sepal_length"], iris["petal_length"])
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()
```

```python
sns.scatterplot(data=iris, x="sepal_length", y="petal_length", hue="species")
plt.show()
```

```python
print(sns.get_dataset_names())
```

```python
iris.info()
```

```python
sns.scatterplot(data=iris, x="sepal_length", y="sepal_width", hue="species")
plt.title("Sepal Length vs Sepal Width")
plt.show()
```

```python
sorted_iris = iris.sort_values("petal_length")
sns.lineplot(data=sorted_iris, x="sepal_length", y="petal_length", hue="species")
plt.show()
```

```python
sns.barplot(data=iris, x="species", y="sepal_length")
plt.show()
```

```python
sns.countplot(data=iris, x="species")
plt.show()
```

```python
sns.histplot(data=iris, x="petal_length", hue="species",bins=20)
plt.show()
```

```python
sns.kdeplot(data=iris, x="petal_length", hue="species", fill=True)
plt.show()
```

```python
sns.violinplot(data=iris, x="species", y="petal_length")
plt.show()
```

```python
sns.swarmplot(data=iris, x="species", y="petal_length", size=4)
plt.show()
```

```python
sns.stripplot(data=iris, x="species", y="petal_length", jitter=True)
plt.show()
```

```python
sns.relplot(data=iris, x="sepal_length", y="petal_length", hue="species", col="species", kind="line")
plt.show()
```

```python
sns.regplot(data=iris, x="sepal_length", y="petal_length")
plt.show()
```

```python
sns.lmplot(data=iris, x="sepal_length", y="petal_length", hue="species")
plt.show()
```

```python
corr = iris.drop(columns="species").corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.show()
```

```python
sns.pairplot(iris, hue="species", diag_kind="kde")
plt.show()
```
